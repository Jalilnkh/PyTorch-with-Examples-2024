## Load dataset and Tokenzire

```python
from datasets import load_dataset
from transformers import T5Tokenizer
# Load your tokenizer
```

```python
from transformers import T5Tokenizer
# Load the tokenizer from Hugging Face Hub
tokenizer = T5Tokenizer.from_pretrained("jalilkartal/AZB_EN_48m")
```

## Add and check special tokens to the tokenizer

```python

# Define the special tokens
special_tokens_dict = {
    "additional_special_tokens": [
        "Translate English to South Azerbaijani:",
        "Translate South Azerbaijani to English:"
    ]
}

# Add special tokens to the tokenizer
num_added_tokens = tokenizer.add_special_tokens(special_tokens_dict)
print(f"Added {num_added_tokens} special tokens.")
```

```python
# Verify the tokens
print("Special tokens:")
print(f"PAD Token: {tokenizer.pad_token}")
print(f"EOS Token: {tokenizer.eos_token}")
print(f"Added Special Tokens: {tokenizer.additional_special_tokens}")
```

```python
print("Special Tokens:", tokenizer.additional_special_tokens)
print("Token IDs:")
print(f"'Translate English to South Azerbaijani:' -> {tokenizer.convert_tokens_to_ids('Translate English to South Azerbaijani:')}")
print(f"'Translate South Azerbaijani to English:' -> {tokenizer.convert_tokens_to_ids('Translate South Azerbaijani to English:')}")
```

## 1. Basic Functionality Test

```python
# Test strings
test_text = "سلام نئجه سن؟ بو بیر تست دیر"

# Tokenize the string
tokens = tokenizer.tokenize(test_text)
print("Tokens:", tokens)

# Encode the string to token IDs
encoded = tokenizer.encode(test_text)
print("Encoded IDs:", encoded)

# Decode back to text
decoded_text = tokenizer.decode(encoded)
print("Decoded Text:", decoded_text)

# Verify round-trip consistency
assert test_text.strip() == decoded_text.strip(), "Decoded text doesn't match the original text!"
```

```python
# Test strings
test_text = "Hi how are you? this is a test."

# Tokenize the string
tokens = tokenizer.tokenize(test_text)
print("Tokens:", tokens)

# Encode the string to token IDs
encoded = tokenizer.encode(test_text)
print("Encoded IDs:", encoded)

# Decode back to text
decoded_text = tokenizer.decode(encoded)
print("Decoded Text:", decoded_text)

# Verify round-trip consistency
assert test_text.strip() == decoded_text.strip(), "Decoded text doesn't match the original text!"
```

## 2. Vocabulary Check

```python
# Check the vocabulary size
print("Vocabulary Size:", len(tokenizer))

# Search for specific tokens
special_tokens = ["و", "ئ", "وْ"]
for token in special_tokens:
    token_id = tokenizer.convert_tokens_to_ids(token)
    print(f"Token: {token}, ID: {token_id}")
```

## 3. Special Tokens Check

```python
# Verify special tokens
print("Pad Token:", tokenizer.pad_token, "ID:", tokenizer.pad_token_id)
print("EOS Token:", tokenizer.eos_token, "ID:", tokenizer.eos_token_id)
print("UNK Token:", tokenizer.unk_token, "ID:", tokenizer.unk_token_id)

# Add missing special tokens if needed
special_tokens_dict = {
    "pad_token": "<pad>",
    "eos_token": "</s>",
    "unk_token": "<unk>"
}
tokenizer.add_special_tokens(special_tokens_dict)
```

## 4. Handling Long Sequences

```python
long_text = " ".join(["word"] * 1024)  # Simulate a long text
encoded = tokenizer(long_text, truncation=True, max_length=512)
print("Truncated Length:", len(encoded["input_ids"]))
```

## 6. Masking Test (For Span Corruption Pretraining)

```python
# Example masked text
masked_text = "This is a <extra_id_0> test."
# Tokenize and encode
encoded = tokenizer(masked_text, return_tensors="pt")
print("Masked Tokens:", encoded)

# Decode back
decoded = tokenizer.decode(encoded["input_ids"][0])
print("Decoded Text with Mask:", decoded)
```

## 7. Check Token Distribution

```python
from collections import Counter

# Tokenize a sample dataset
sample_texts = ["یوْخلانیلیبدیر", "قایناق‌لار", "دينی اينانجلارينا گؤره"]
tokenized_texts = [tokenizer.tokenize(text) for text in sample_texts]

# Flatten token list and count
flat_tokens = [token for sublist in tokenized_texts for token in sublist]
token_counts = Counter(flat_tokens)
print("Token Distribution:", token_counts)
```

## 8. Validate Vocabulary Coverage

```python
dataset = ["یوْخلانیلیبدیر", "قایناق‌لار", "دينی اينانجلارينا گؤره"]  # Sample South Azerbaijani text

# Check vocabulary coverage
missing_tokens = []
for text in dataset:
    tokens = tokenizer.tokenize(text)
    for token in tokens:
        if token not in tokenizer.get_vocab():
            missing_tokens.append(token)

print("Missing Tokens:", set(missing_tokens))
```
